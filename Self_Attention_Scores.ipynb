```python
import numpy as np
sentence = "What are the symptoms of diabetes"
tokens   = sentence.split()
print(f"Tokens: {tokens}")
```

```python
np.random.seed(42)
d_model = 8
d_k     = 4
```

```python
X = np.random.randn(len(tokens), d_model)
```

```python
W_Q = np.random.randn(d_model, d_k)
W_K = np.random.randn(d_model, d_k)
W_V = np.random.randn(d_model, d_k)
```

```python
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

scores = Q @ K.T / np.sqrt(d_k)
```

```python
def softmax(x):
    e = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e / e.sum(axis=-1, keepdims=True)
attention_weights = softmax(scores)
output = attention_weights @ V
```

```python

print("\n Raw Attention Scores (QKᵀ / √d_k) ")
print(np.round(scores, 3))

print("\n Attention Weights after Softmax ")
print(np.round(attention_weights, 3))

print("\n Self-Attention Output (weighted V) ")
print(np.round(output, 3))
```

```python
print("\n Top attended token per word ")
for i, token in enumerate(tokens):
    top_j = np.argmax(attention_weights[i])
    print(f"  '{token}' attends most to '{tokens[top_j]}'",
          f"(weight={attention_weights[i, top_j]:.3f})")
```
